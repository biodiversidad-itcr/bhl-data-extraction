###This cell is for selecting extracted texts from the BHL API.
#Rows per person:
###Raquel 0-300
###Pamela 301-600
###Manfred 601-1000
###Alejandro 1001-1400


```python
import pandas as pd
import textwrap
import ipywidgets as widgets
from IPython.display import display, clear_output


df = pd.read_csv("RawTexts.csv")
new_df = pd.DataFrame(columns=df.columns)

try:
    start_row = int(input(f"Enter the row number to start from (0 - {len(df)-1}): ").strip())
    if not 0 <= start_row < len(df):
        print("Row number out of range.")
        exit()
except ValueError:
    print("Invalid input. Must be an integer.")
    exit()

current_index = start_row
handling_event = False  # control flag

def show_row():
    clear_output(wait=True)
    if current_index >= len(df):
        print("✅ Finished!")
        print(f"📊 Total rows saved: {len(new_df)}")
        return

    if len(new_df) >= 100:
        print("🎯 Reached the 100-row limit.")
        print(f"📊 Total rows saved: {len(new_df)}")
        return

    row = df.iloc[current_index]
    print(f"\n--- Row {current_index} ---")
    print("Text:\n", textwrap.fill(row["Text"], width=80))
    print(f"✅ Current saved rows: {len(new_df)}")

    if current_index == 300:
        print("📣 Notice: You are at row 300. You are entering in the designed rows of a teammate")
    elif current_index == 600:
        print("📣 Notice: You are at row 600. You are entering in the designed rows of a teammate")

    display(dropdown)

def on_choice(change):
    global current_index, new_df, handling_event
    if handling_event or change['new'] is None:
        return
    handling_event = True

    choice = change['new']
    if choice == 'y':
        new_df = pd.concat([new_df, pd.DataFrame([df.iloc[current_index]])], ignore_index=True)
    elif choice == 'exit':
        clear_output(wait=True)
        print("Exiting early.")
        print("You finished at row:", current_index)
        print(f"📊 Total rows saved: {len(new_df)}")
        handling_event = False
        return
    current_index += 1
    dropdown.value = None
    handling_event = False
    show_row()

dropdown = widgets.Dropdown(
    options=[('Save', 'y'), ('Skip', 'n'), ('Exit', 'exit')],
    description="Save this row?",
    value=None,
    style={'description_width': 'initial'}
)

dropdown.observe(on_choice, names='value')
show_row()
```
